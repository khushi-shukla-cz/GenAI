```python

# ============================================================
# GENERATIVE AI LABORATORY - EXPERIMENT 6
# Generative Adversarial Network (GAN) for Synthetic Image Generation
# Dataset: MNIST
# Framework: TensorFlow / Keras
# ============================================================

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import os
import shutil

from tensorflow.keras import layers, Model
from tensorflow.keras.datasets import mnist

# Set random seeds
tf.random.set_seed(42)
np.random.seed(42)

# Create output directory
os.makedirs("gan_results", exist_ok=True)

# ------------------------------------------------------------
# 1. LOAD AND PREPROCESS MNIST DATASET
# ------------------------------------------------------------

(x_train, _), (x_test, _) = mnist.load_data()

# Normalize pixel values to [-1, 1]
x_train = (x_train.astype("float32") - 127.5) / 127.5
x_test = (x_test.astype("float32") - 127.5) / 127.5

# Reshape images to 28 x 28 x 1
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

print("Training data shape:", x_train.shape)
print("Testing data shape:", x_test.shape)

# Display real MNIST images
plt.figure(figsize=(10, 4))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i].squeeze(), cmap="gray")
    plt.axis("off")
plt.suptitle("Real MNIST Images")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 2. BUILD GENERATOR
# ------------------------------------------------------------

latent_dim = 100

def build_generator():
    noise = layers.Input(shape=(latent_dim,))

    x = layers.Dense(256)(noise)
    x = layers.LeakyReLU(negative_slope=0.2)(x)

    x = layers.Dense(512)(x)
    x = layers.LeakyReLU(negative_slope=0.2)(x)

    x = layers.Dense(1024)(x)
    x = layers.LeakyReLU(negative_slope=0.2)(x)

    x = layers.Dense(28 * 28, activation="tanh")(x)
    image = layers.Reshape((28, 28, 1))(x)

    return Model(noise, image, name="Generator")

generator = build_generator()

print("\nGENERATOR MODEL SUMMARY")
generator.summary()

# ------------------------------------------------------------
# 3. BUILD DISCRIMINATOR
# ------------------------------------------------------------

def build_discriminator():
    image = layers.Input(shape=(28, 28, 1))

    x = layers.Flatten()(image)

    x = layers.Dense(512)(x)
    x = layers.LeakyReLU(negative_slope=0.2)(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Dense(256)(x)
    x = layers.LeakyReLU(negative_slope=0.2)(x)
    x = layers.Dropout(0.3)(x)

    output = layers.Dense(1, activation="sigmoid")(x)

    return Model(image, output, name="Discriminator")

discriminator = build_discriminator()

print("\nDISCRIMINATOR MODEL SUMMARY")
discriminator.summary()

# ------------------------------------------------------------
# 4. DEFINE LOSS FUNCTION AND OPTIMIZERS
# ------------------------------------------------------------

bce = tf.keras.losses.BinaryCrossentropy()

d_optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.0002, beta_1=0.5
)

g_optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.0002, beta_1=0.5
)

# ------------------------------------------------------------
# 5. CREATE GAN TRAINING STEP
# ------------------------------------------------------------

@tf.function
def train_step(real_images):
    current_batch_size = tf.shape(real_images)[0]

    # Train Discriminator
    noise = tf.random.normal(
        (current_batch_size, latent_dim)
    )

    with tf.GradientTape() as d_tape:
        fake_images = generator(noise, training=True)

        real_output = discriminator(
            real_images, training=True
        )

        fake_output = discriminator(
            tf.stop_gradient(fake_images), training=True
        )

        d_loss_real = bce(
            tf.ones_like(real_output), real_output
        )

        d_loss_fake = bce(
            tf.zeros_like(fake_output), fake_output
        )

        d_loss = (d_loss_real + d_loss_fake) / 2

    d_gradients = d_tape.gradient(
        d_loss, discriminator.trainable_variables
    )

    d_optimizer.apply_gradients(
        zip(d_gradients, discriminator.trainable_variables)
    )

    # Train Generator to fool Discriminator
    noise = tf.random.normal(
        (current_batch_size, latent_dim)
    )

    with tf.GradientTape() as g_tape:
        fake_images = generator(noise, training=True)

        fake_output = discriminator(
            fake_images, training=False
        )

        g_loss = bce(
            tf.ones_like(fake_output), fake_output
        )

    g_gradients = g_tape.gradient(
        g_loss, generator.trainable_variables
    )

    g_optimizer.apply_gradients(
        zip(g_gradients, generator.trainable_variables)
    )

    return d_loss, g_loss

# ------------------------------------------------------------
# 6. GENERATE AND SAVE SYNTHETIC IMAGES
# ------------------------------------------------------------

fixed_noise = tf.random.normal((25, latent_dim))

def generate_images(epoch):
    generated = generator(
        fixed_noise, training=False
    ).numpy()

    plt.figure(figsize=(8, 8))

    for i in range(25):
        plt.subplot(5, 5, i + 1)

        image = (generated[i].squeeze() + 1) / 2

        plt.imshow(
            image, cmap="gray", vmin=0, vmax=1
        )
        plt.axis("off")

    plt.suptitle(f"Generated Images - Epoch {epoch}")
    plt.tight_layout()

    plt.savefig(
        f"gan_results/epoch_{epoch:03d}.png",
        dpi=150,
        bbox_inches="tight"
    )

    plt.show()

# ------------------------------------------------------------
# 7. TRAIN GAN FOR 30 EPOCHS
# ------------------------------------------------------------

epochs = 30
batch_size = 128

generator_losses = []
discriminator_losses = []

# Display initial generated images
generate_images(0)

for epoch in range(1, epochs + 1):

    d_epoch_losses = []
    g_epoch_losses = []

    # Shuffle dataset
    indices = np.random.permutation(len(x_train))
    shuffled_images = x_train[indices]

    # Mini-batch training
    for start in range(0, len(x_train) - batch_size + 1, batch_size):

        real_batch = shuffled_images[
            start:start + batch_size
        ]

        d_loss, g_loss = train_step(
            tf.convert_to_tensor(real_batch)
        )

        d_epoch_losses.append(float(d_loss.numpy()))
        g_epoch_losses.append(float(g_loss.numpy()))

    avg_d_loss = np.mean(d_epoch_losses)
    avg_g_loss = np.mean(g_epoch_losses)

    discriminator_losses.append(avg_d_loss)
    generator_losses.append(avg_g_loss)

    print(
        f"Epoch {epoch:02d}/{epochs} | "
        f"Discriminator Loss: {avg_d_loss:.4f} | "
        f"Generator Loss: {avg_g_loss:.4f}"
    )

    # Save images at selected epochs
    if epoch in [1, 5, 10, 20, 30]:
        generate_images(epoch)

print("\nGAN training completed successfully!")

# ------------------------------------------------------------
# 8. PLOT GENERATOR AND DISCRIMINATOR LOSSES
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    range(1, epochs + 1),
    generator_losses,
    label="Generator Loss"
)

plt.plot(
    range(1, epochs + 1),
    discriminator_losses,
    label="Discriminator Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Binary Cross-Entropy Loss")
plt.title("GAN Training Loss")
plt.legend()
plt.grid(True)

plt.savefig(
    "gan_results/loss_graph.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------
# 9. COMPARE REAL AND SYNTHETIC IMAGES
# ------------------------------------------------------------

noise = tf.random.normal((10, latent_dim))

synthetic_images = generator(
    noise, training=False
).numpy()

plt.figure(figsize=(12, 5))

for i in range(10):

    # Real images
    plt.subplot(2, 10, i + 1)
    plt.imshow(
        x_test[i].squeeze(), cmap="gray",
        vmin=-1, vmax=1
    )
    plt.axis("off")

    # Synthetic images
    plt.subplot(2, 10, i + 11)
    plt.imshow(
        (synthetic_images[i].squeeze() + 1) / 2,
        cmap="gray", vmin=0, vmax=1
    )
    plt.axis("off")

plt.figtext(
    0.25, 0.92, "Real Images",
    ha="center", fontsize=12
)

plt.figtext(
    0.72, 0.92, "Synthetic Images",
    ha="center", fontsize=12
)

plt.suptitle("Real vs Synthetic MNIST Images")
plt.tight_layout(rect=[0, 0, 1, 0.88])

plt.savefig(
    "gan_results/real_vs_synthetic.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------
# 10. SAVE TRAINED MODELS
# ------------------------------------------------------------

generator.save("gan_results/generator.keras")
discriminator.save("gan_results/discriminator.keras")

print("Generator model saved.")
print("Discriminator model saved.")

# ------------------------------------------------------------
# 11. SAVE LOSS VALUES
# ------------------------------------------------------------

np.savetxt(
    "gan_results/generator_losses.csv",
    generator_losses,
    delimiter=",",
    header="Generator Loss",
    comments=""
)

np.savetxt(
    "gan_results/discriminator_losses.csv",
    discriminator_losses,
    delimiter=",",
    header="Discriminator Loss",
    comments=""
)

# ------------------------------------------------------------
# 12. ZIP ALL RESULTS FOR DOWNLOAD
# ------------------------------------------------------------

shutil.make_archive(
    "GAN_Experiment_6_Results",
    "zip",
    "gan_results"
)

print("\nAll experiment results saved successfully.")
print("ZIP file: GAN_Experiment_6_Results.zip")

try:
    from google.colab import files
    files.download("GAN_Experiment_6_Results.zip")
except ImportError:
    print("Run this notebook in Google Colab to download the ZIP automatically.")

# ------------------------------------------------------------
# END OF EXPERIMENT 6
# ------------------------------------------------------------
```
