```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

import re
import matplotlib.pyplot as plt
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from wordcloud import WordCloud
from gensim.models import Word2Vec
from collections import Counter
```

```python
text = \"\"\"
Artificial intelligence is transforming the way people learn, work, communicate,
and solve problems. Generative AI systems can create text, summarize documents,
answer questions, generate code, and support students in their educational
activities. Natural Language Processing is an important field of artificial
intelligence that helps computers understand and process human language.
Before textual information can be given to a machine learning model, it usually
needs to be cleaned and transformed into a suitable format. Text preprocessing
helps remove unnecessary information and makes the data more consistent.
Common preprocessing techniques include converting text into lowercase, removing
punctuation, handling special characters, removing stop words, and correcting
or standardizing words.
Tokenization is another important step in Natural Language Processing. It divides
a sentence or paragraph into smaller units called tokens. These tokens may be
individual words, subwords, or characters. After tokenization, stop words such as
the, is, are, and of can be removed when they do not provide useful information.
Stemming reduces words to a basic root form by removing prefixes or suffixes.
Lemmatization performs a similar task but attempts to return a meaningful
dictionary form of a word. For example, words such as running, runs, and ran may
be related to the lemma run.
Word embeddings convert words into numerical vectors. These vectors allow
machine learning systems to represent relationships between words. Word2Vec is
a popular technique that learns word representations from the context in which
words appear. Similar words can have similar vector representations.
Modern transformer models also use tokenization and numerical representations
to process language. BERT and GPT use subword-based approaches to handle words
and unfamiliar terms efficiently. Good preprocessing can improve the quality of
many NLP applications including sentiment analysis, classification, summarization,
question answering, information retrieval, chatbots, and text generation.
Therefore, preprocessing and representation of text are important stages in
building effective Natural Language Processing systems.
\"\"\"

print("ORIGINAL TEXT:\\n", text)
print("Word count:", len(text.split()))
```

```python
lower_text = text.lower()
print("LOWERCASE TEXT:\\n", lower_text)
```

```python
clean_text = re.sub(r'[^a-zA-Z\\s]', '', lower_text)
print("PREPROCESSED TEXT:\\n", clean_text)
```

```python
tokens = word_tokenize(clean_text)
print("TOKENS:")
print(tokens)
print("Total number of tokens:", len(tokens))
```

```python
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word not in stop_words]
print("TOKENS AFTER STOP WORD REMOVAL:")
print(filtered_tokens)
print("Number of tokens after stop-word removal:", len(filtered_tokens))
```

```python
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_tokens]
print("STEMMED WORDS:")
print(stemmed_words)
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_tokens]
print("LEMMATIZED WORDS:")
print(lemmatized_words)
```

```python
word_frequency = Counter(filtered_tokens)
print("WORD FREQUENCY:")
for word, count in word_frequency.most_common(20):
    print(word, ":", count)
```

```python
wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color="white"
).generate(" ".join(filtered_tokens))

plt.figure(figsize=(15, 7))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud of Preprocessed Text")
plt.show()
```

```python
sentences = [
    word_tokenize(re.sub(r'[^a-zA-Z\\s]', '', sentence.lower()))
    for sentence in text.split('.')
]
sentences = [
    [word for word in sentence if word not in stop_words]
    for sentence in sentences
    if sentence
]

print("Sentences prepared for Word2Vec:")
print(sentences)

model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    seed=42
)

print("Word2Vec model trained successfully.")
print("Vocabulary:")
print(list(model.wv.index_to_key))
```

```python
selected_words = [
    "artificial",
    "intelligence",
    "language",
    "processing",
    "learning"
]

for word in selected_words:
    if word in model.wv:
        print("\\nWord:", word)
        print("Embedding Vector:")
        print(model.wv[word])
    else:
        print("\\nWord not found in vocabulary:", word)
```

```python
keyword = "language"

if keyword in model.wv:
    similar_words = model.wv.most_similar(keyword, topn=3)
    print("Three words most similar to:", keyword)
    for word, similarity in similar_words:
        print(word, ":", similarity)
else:
    print("Keyword not found in vocabulary.")
```

```python
print("=" * 60)
print("TEXT PREPROCESSING AND WORD EMBEDDING RESULTS")
print("=" * 60)

print("\\n1. ORIGINAL TEXT:")
print(text[:500], "...")

print("\\n2. NUMBER OF TOKENS:")
print(len(tokens))

print("\\n3. TOKENS AFTER STOP-WORD REMOVAL:")
print(filtered_tokens[:30])

print("\\n4. STEMMED WORDS:")
print(stemmed_words[:30])

print("\\n5. LEMMATIZED WORDS:")
print(lemmatized_words[:30])

print("\\n6. TOP 10 WORD FREQUENCIES:")
for word, count in word_frequency.most_common(10):
    print(word, ":", count)

print("\\n7. WORD2VEC MODEL:")
print("Model trained successfully.")

print("\\n8. SELECTED WORD EMBEDDINGS (first 10 values):")
for word in selected_words:
    if word in model.wv:
        print(word, "->", model.wv[word][:10], "...")

print("\\n9. THREE MOST SIMILAR WORDS TO", keyword)
if keyword in model.wv:
    for word, similarity in model.wv.most_similar(keyword, topn=3):
        print(word, ":", similarity)

print("\\nEXPERIMENT COMPLETED SUCCESSFULLY!")
```
