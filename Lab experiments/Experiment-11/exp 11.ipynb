```python
from transformers import pipeline
print("Transformers library imported successfully!")
```

```python
generator = pipeline("text-generation", model="gpt2")
print("GPT-2 text generation model loaded successfully!")
```

```python
prompt = "Generative AI is transforming education because"

output = generator(
    prompt,
    max_new_tokens=60,
    num_return_sequences=1
)

print(output[0]["generated_text"])
```

```python
prompts = [
    "Artificial Intelligence will shape the future of",
    "Generative AI can help students by",
    "Python programming is useful because",
    "Technology can improve healthcare by",
    "One limitation of artificial intelligence is"
]

for i, prompt in enumerate(prompts, start=1):
    output = generator(
        prompt,
        max_new_tokens=60,
        num_return_sequences=1
    )
    print(f"Experiment {i}")
    print("Prompt:", prompt)
    print("Generated text:", output[0]["generated_text"])
    print("-" * 50)
```

```python
prompt = "Artificial Intelligence will shape the future of"

# Test A: Default generation settings
result_a = generator(
    prompt,
    max_new_tokens=60,
    num_return_sequences=1
)

# Test B: Sampling enabled
result_b = generator(
    prompt,
    max_new_tokens=60,
    do_sample=True,
    temperature=0.8,
    top_k=50,
    num_return_sequences=1
)

# Test C: Lower temperature
result_c = generator(
    prompt,
    max_new_tokens=60,
    do_sample=True,
    temperature=0.3,
    top_k=50,
    num_return_sequences=1
)

# Test D: Higher temperature
result_d = generator(
    prompt,
    max_new_tokens=60,
    do_sample=True,
    temperature=1.2,
    top_k=50,
    num_return_sequences=1
)

# Test E: Change top_k
result_e = generator(
    prompt,
    max_new_tokens=60,
    do_sample=True,
    temperature=0.8,
    top_k=10,
    num_return_sequences=1
)

tests = [
    ("Default settings", result_a),
    ("Temperature 0.8, top_k 50", result_b),
    ("Temperature 0.3, top_k 50", result_c),
    ("Temperature 1.2, top_k 50", result_d),
    ("Temperature 0.8, top_k 10", result_e)
]

for name, result in tests:
    print("\n", name)
    print(result[0]["generated_text"])
    print("-" * 50)
```
