```python
# ============================================================
# EXPERIMENT 4
# Variational Autoencoder (VAE) for Image Generation,
# Reconstruction, Latent Space Visualization, and Evaluation
# Dataset: MNIST Handwritten Digits
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------
import tensorflow as tf
from tensorflow.keras import layers, Model
from tensorflow.keras.datasets import mnist
import numpy as np
import matplotlib.pyplot as plt

print("TensorFlow Version:", tf.__version__)

# Set seeds for reproducible results.
np.random.seed(42)
tf.random.set_seed(42)


# ------------------------------------------------------------
# 2. LOAD THE MNIST DATASET
# ------------------------------------------------------------
(x_train, y_train), (x_test, y_test) = mnist.load_data()

print("Training data shape:", x_train.shape)
print("Testing data shape :", x_test.shape)


# ------------------------------------------------------------
# 3. NORMALIZE AND RESHAPE IMAGES
# ------------------------------------------------------------
# Normalize pixel values from [0, 255] to [0, 1].
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Flatten 28 x 28 images into 784-dimensional vectors.
x_train = x_train.reshape((-1, 784))
x_test = x_test.reshape((-1, 784))

print("After reshaping:")
print("x_train:", x_train.shape)
print("x_test :", x_test.shape)


# ------------------------------------------------------------
# 4. DEFINE THE LATENT DIMENSION
# ------------------------------------------------------------
# A 2-D latent space allows us to visualize encoded digits.
latent_dim = 2
print("Latent dimension:", latent_dim)


# ------------------------------------------------------------
# 5. BUILD THE ENCODER
# ------------------------------------------------------------
encoder_inputs = layers.Input(shape=(784,), name="encoder_input")

x = layers.Dense(256, activation="relu")(encoder_inputs)
x = layers.Dense(128, activation="relu")(x)

# Learn the mean and log variance of the latent distribution.
z_mean = layers.Dense(latent_dim, name="z_mean")(x)
z_log_var = layers.Dense(latent_dim, name="z_log_var")(x)


# ------------------------------------------------------------
# 6. REPARAMETERIZATION / SAMPLING LAYER
# ------------------------------------------------------------
class Sampling(layers.Layer):
    """Sample latent vectors using the reparameterization trick."""

    def call(self, inputs):
        z_mean, z_log_var = inputs

        batch_size = tf.shape(z_mean)[0]
        dimension = tf.shape(z_mean)[1]

        # Draw random noise from a standard normal distribution.
        epsilon = tf.random.normal(
            shape=(batch_size, dimension)
        )

        # Convert the learned distribution into a latent sample.
        return z_mean + tf.exp(0.5 * z_log_var) * epsilon


z = Sampling(name="z")([z_mean, z_log_var])

encoder = Model(
    encoder_inputs,
    [z_mean, z_log_var, z],
    name="encoder"
)

print("\nEncoder Architecture:")
encoder.summary()


# ------------------------------------------------------------
# 7. BUILD THE DECODER
# ------------------------------------------------------------
latent_inputs = layers.Input(
    shape=(latent_dim,),
    name="z_sampling"
)

x = layers.Dense(128, activation="relu")(latent_inputs)
x = layers.Dense(256, activation="relu")(x)

# Sigmoid keeps reconstructed pixel values between 0 and 1.
decoder_outputs = layers.Dense(
    784,
    activation="sigmoid"
)(x)

decoder = Model(
    latent_inputs,
    decoder_outputs,
    name="decoder"
)

print("\nDecoder Architecture:")
decoder.summary()


# ------------------------------------------------------------
# 8. DEFINE THE VAE MODEL AND LOSS FUNCTIONS
# ------------------------------------------------------------
class VAE(Model):

    def __init__(self, encoder, decoder, **kwargs):
        super().__init__(**kwargs)

        self.encoder = encoder
        self.decoder = decoder

        self.total_loss_tracker = tf.keras.metrics.Mean(
            name="loss"
        )

        self.reconstruction_loss_tracker = tf.keras.metrics.Mean(
            name="reconstruction_loss"
        )

        self.kl_loss_tracker = tf.keras.metrics.Mean(
            name="kl_loss"
        )

    @property
    def metrics(self):
        # Keras resets these metrics at the beginning of each epoch.
        return [
            self.total_loss_tracker,
            self.reconstruction_loss_tracker,
            self.kl_loss_tracker
        ]

    def compute_vae_losses(self, data, training=False):

        # Encode the input into a latent distribution and sample.
        z_mean, z_log_var, z = self.encoder(
            data,
            training=training
        )

        # Decode the latent sample to reconstruct the image.
        reconstruction = self.decoder(
            z,
            training=training
        )

        # Reconstruction loss: binary cross-entropy over pixels.
        pixel_bce = tf.keras.backend.binary_crossentropy(
            data,
            reconstruction
        )

        reconstruction_loss = tf.reduce_mean(
            tf.reduce_sum(pixel_bce, axis=1)
        )

        # KL divergence regularizes the latent distribution
        # toward a standard normal distribution N(0, I).
        kl_per_sample = -0.5 * tf.reduce_sum(
            1.0
            + z_log_var
            - tf.square(z_mean)
            - tf.exp(z_log_var),
            axis=1
        )

        kl_loss = tf.reduce_mean(kl_per_sample)

        # Total VAE loss combines reconstruction and KL losses.
        total_loss = reconstruction_loss + kl_loss

        return total_loss, reconstruction_loss, kl_loss

    def train_step(self, data):

        # Extract the input tensor if Keras supplies a tuple.
        if isinstance(data, (tuple, list)):
            data = data[0]

        with tf.GradientTape() as tape:
            total_loss, reconstruction_loss, kl_loss = (
                self.compute_vae_losses(data, training=True)
            )

        # Compute and apply gradients to train the network.
        gradients = tape.gradient(
            total_loss,
            self.trainable_weights
        )

        self.optimizer.apply_gradients(
            zip(gradients, self.trainable_weights)
        )

        self.total_loss_tracker.update_state(total_loss)
        self.reconstruction_loss_tracker.update_state(
            reconstruction_loss
        )
        self.kl_loss_tracker.update_state(kl_loss)

        return {
            metric.name: metric.result()
            for metric in self.metrics
        }

    def test_step(self, data):
        # Compute validation losses without updating weights.
        if isinstance(data, (tuple, list)):
            data = data[0]

        total_loss, reconstruction_loss, kl_loss = (
            self.compute_vae_losses(data, training=False)
        )

        self.total_loss_tracker.update_state(total_loss)
        self.reconstruction_loss_tracker.update_state(
            reconstruction_loss
        )
        self.kl_loss_tracker.update_state(kl_loss)

        return {
            metric.name: metric.result()
            for metric in self.metrics
        }


# ------------------------------------------------------------
# 9. CREATE AND COMPILE THE VAE
# ------------------------------------------------------------
vae = VAE(encoder, decoder, name="vae")

vae.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    )
)


# ------------------------------------------------------------
# 10. TRAIN THE VAE
# ------------------------------------------------------------
epochs = 20
batch_size = 128

history = vae.fit(
    x_train,
    epochs=epochs,
    batch_size=batch_size,
    validation_data=(x_test, None),
    verbose=1
)


# ------------------------------------------------------------
# 11. PLOT TRAINING AND VALIDATION LOSS
# ------------------------------------------------------------
plt.figure(figsize=(9, 5))

plt.plot(
    history.history["loss"],
    label="Training Total Loss"
)

plt.plot(
    history.history["reconstruction_loss"],
    label="Training Reconstruction Loss"
)

plt.plot(
    history.history["kl_loss"],
    label="Training KL Loss"
)

if "val_loss" in history.history:
    plt.plot(
        history.history["val_loss"],
        "--",
        label="Validation Total Loss"
    )

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("VAE Training and Validation Loss")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 12. RECONSTRUCT TEST IMAGES
# ------------------------------------------------------------
# Encode ten test images and obtain their latent samples.
z_mean_test, z_log_var_test, z_test = encoder.predict(
    x_test[:10],
    verbose=0
)

# Decode latent samples to reconstruct the original images.
reconstructed_images = decoder.predict(
    z_test,
    verbose=0
)


# ------------------------------------------------------------
# 13. DISPLAY ORIGINAL VS. RECONSTRUCTED IMAGES
# ------------------------------------------------------------
plt.figure(figsize=(12, 4))

for i in range(10):

    # First row: original images.
    ax = plt.subplot(2, 10, i + 1)

    plt.imshow(
        x_test[i].reshape(28, 28),
        cmap="gray"
    )

    plt.axis("off")

    if i == 0:
        ax.set_title("Original")

    # Second row: reconstructed images.
    ax = plt.subplot(2, 10, i + 11)

    plt.imshow(
        reconstructed_images[i].reshape(28, 28),
        cmap="gray"
    )

    plt.axis("off")

    if i == 0:
        ax.set_title("Reconstructed")

plt.suptitle("Original vs. Reconstructed MNIST Images")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 14. GENERATE 10 NEW IMAGES
# ------------------------------------------------------------
num_generated = 10

# Sample random points from the standard normal latent space.
random_latent_vectors = np.random.normal(
    size=(num_generated, latent_dim)
).astype("float32")

generated_images = decoder.predict(
    random_latent_vectors,
    verbose=0
)


# ------------------------------------------------------------
# 15. DISPLAY THE 10 GENERATED IMAGES
# ------------------------------------------------------------
plt.figure(figsize=(12, 3))

for i in range(num_generated):
    plt.subplot(1, num_generated, i + 1)

    plt.imshow(
        generated_images[i].reshape(28, 28),
        cmap="gray"
    )

    plt.axis("off")

plt.suptitle("Images Generated from Random Latent Vectors")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 16. VISUALIZE THE 2-D LATENT SPACE
# ------------------------------------------------------------
# Encode the entire test set and plot the mean latent vectors.
z_mean_all, z_log_var_all, z_all = encoder.predict(
    x_test,
    batch_size=256,
    verbose=0
)

plt.figure(figsize=(9, 7))

scatter = plt.scatter(
    z_mean_all[:, 0],
    z_mean_all[:, 1],
    c=y_test,
    cmap="tab10",
    alpha=0.6,
    s=8
)

plt.colorbar(
    scatter,
    ticks=range(10),
    label="Digit"
)

plt.xlabel("Latent Dimension 1")
plt.ylabel("Latent Dimension 2")
plt.title("2-D Latent Space Visualization of MNIST")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 17. DISPLAY THE VAE ARCHITECTURE
# ------------------------------------------------------------
print("\nEncoder Architecture:")
encoder.summary()

print("\nDecoder Architecture:")
decoder.summary()


# ============================================================
# ADDITIONAL OUTPUTS
# ============================================================


# ------------------------------------------------------------
# 18. GENERATE A 5 x 5 GRID OF 25 NEW IMAGES
# ------------------------------------------------------------
num_generated_grid = 25

latent_grid = np.random.normal(
    size=(num_generated_grid, latent_dim)
).astype("float32")

generated_grid_images = decoder.predict(
    latent_grid,
    verbose=0
)

plt.figure(figsize=(7, 7))

for i in range(num_generated_grid):
    plt.subplot(5, 5, i + 1)

    plt.imshow(
        generated_grid_images[i].reshape(28, 28),
        cmap="gray"
    )

    plt.axis("off")

plt.suptitle("25 Images Generated by the VAE", fontsize=14)
plt.tight_layout()
plt.show()

print("Images generated in the 5 x 5 grid:", num_generated_grid)


# ------------------------------------------------------------
# 19. LATENT SPACE INTERPOLATION
# ------------------------------------------------------------
# Select two test images and interpolate between their
# encoder mean vectors to visualize a gradual transition.
start_index = 0
end_index = 7

z_mean_pair, _, _ = encoder.predict(
    x_test[[start_index, end_index]],
    verbose=0
)

z_start = z_mean_pair[0]
z_end = z_mean_pair[1]

steps = 10
alphas = np.linspace(0.0, 1.0, steps)

# Generate intermediate latent vectors between the endpoints.
interpolated_latents = np.array([
    (1.0 - alpha) * z_start + alpha * z_end
    for alpha in alphas
], dtype="float32")

interpolated_images = decoder.predict(
    interpolated_latents,
    verbose=0
)

plt.figure(figsize=(15, 3))

for i, image in enumerate(interpolated_images):
    plt.subplot(1, steps, i + 1)

    plt.imshow(
        image.reshape(28, 28),
        cmap="gray"
    )

    plt.title(f"t={alphas[i]:.1f}", fontsize=9)
    plt.axis("off")

plt.suptitle(
    f"Latent Interpolation: Image {start_index} "
    f"(digit {y_test[start_index]}) to Image {end_index} "
    f"(digit {y_test[end_index]})"
)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 20. RECONSTRUCTION ERROR ANALYSIS
# ------------------------------------------------------------
# Measure reconstruction errors for the first 100 test images.
num_error_samples = 100

original_error_images = x_test[:num_error_samples]

_, _, latent_error = encoder.predict(
    original_error_images,
    verbose=0
)

reconstructed_error_images = decoder.predict(
    latent_error,
    verbose=0
)

# Calculate one mean squared error (MSE) value per image.
mse_values = np.mean(
    np.square(
        original_error_images - reconstructed_error_images
    ),
    axis=1
)

print("\nReconstruction Error Statistics")
print("Mean Reconstruction MSE:",
      round(float(np.mean(mse_values)), 6))

print("Minimum Reconstruction MSE:",
      round(float(np.min(mse_values)), 6))

print("Maximum Reconstruction MSE:",
      round(float(np.max(mse_values)), 6))

plt.figure(figsize=(9, 5))

plt.hist(
    mse_values,
    bins=20,
    edgecolor="black"
)

plt.xlabel("Mean Squared Error (MSE)")
plt.ylabel("Number of Images")
plt.title("Distribution of VAE Reconstruction Errors")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 21. RECONSTRUCTION QUALITY WITH MSE VALUES
# ------------------------------------------------------------
num_comparison_samples = 8

original_comparison = x_test[:num_comparison_samples]

_, _, comparison_latents = encoder.predict(
    original_comparison,
    verbose=0
)

reconstructed_comparison = decoder.predict(
    comparison_latents,
    verbose=0
)

plt.figure(figsize=(14, 5))

for i in range(num_comparison_samples):

    # Display the original image.
    plt.subplot(2, num_comparison_samples, i + 1)

    plt.imshow(
        original_comparison[i].reshape(28, 28),
        cmap="gray"
    )

    plt.title("Original")
    plt.axis("off")

    # Display its reconstruction and calculate its MSE.
    plt.subplot(
        2,
        num_comparison_samples,
        num_comparison_samples + i + 1
    )

    plt.imshow(
        reconstructed_comparison[i].reshape(28, 28),
        cmap="gray"
    )

    image_mse = np.mean(
        np.square(
            original_comparison[i] - reconstructed_comparison[i]
        )
    )

    plt.title(f"MSE: {image_mse:.4f}")
    plt.axis("off")

plt.suptitle("Original and Reconstructed Digits with MSE")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 22. GENERATED PIXEL INTENSITY DISTRIBUTION
# ------------------------------------------------------------
# Examine the pixel-value distribution of the 25 generated images.
generated_pixels = generated_grid_images.flatten()

plt.figure(figsize=(8, 5))

plt.hist(
    generated_pixels,
    bins=30,
    edgecolor="black"
)

plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.title("Pixel Intensity Distribution of Generated Images")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

print(
    "Generated pixel intensity range:",
    round(float(generated_pixels.min()), 4),
    "to",
    round(float(generated_pixels.max()), 4)
)


# ------------------------------------------------------------
# 23. LATENT SPACE BY DIGIT CLASS WITH A LEGEND
# ------------------------------------------------------------
# Plot each digit class separately to compare their distributions.
plt.figure(figsize=(9, 7))

for digit in range(10):
    mask = y_test == digit

    plt.scatter(
        z_mean_all[mask, 0],
        z_mean_all[mask, 1],
        s=8,
        alpha=0.55,
        label=str(digit)
    )

plt.xlabel("Latent Dimension 1")
plt.ylabel("Latent Dimension 2")
plt.title("Latent Space Distribution by Digit Class")
plt.legend(title="Digit", markerscale=2, ncol=2)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 24. QUANTITATIVE PERFORMANCE SUMMARY
# ------------------------------------------------------------
# Evaluate the trained VAE on the held-out test dataset.
test_metrics = vae.evaluate(
    x_test,
    batch_size=batch_size,
    verbose=0,
    return_dict=True
)

# Decode the latent samples for test-set reconstruction MSE.
_, _, z_test_all = encoder.predict(
    x_test,
    batch_size=256,
    verbose=0
)

test_reconstructions = decoder.predict(
    z_test_all,
    batch_size=256,
    verbose=0
)

test_mse_values = np.mean(
    np.square(x_test - test_reconstructions),
    axis=1
)

print("\n========== VAE PERFORMANCE SUMMARY ==========")
print("Dataset: MNIST")
print("Training images:", x_train.shape[0])
print("Test images:", x_test.shape[0])
print("Latent dimension:", latent_dim)
print("Epochs trained:", epochs)
print("Batch size:", batch_size)

print(
    "Test total loss:",
    round(float(test_metrics["loss"]), 4)
)

print(
    "Test reconstruction loss:",
    round(float(test_metrics["reconstruction_loss"]), 4)
)

print(
    "Test KL divergence loss:",
    round(float(test_metrics["kl_loss"]), 4)
)

print(
    "Mean test reconstruction MSE:",
    round(float(np.mean(test_mse_values)), 6)
)
›
print("Number of generated images in grid:", num_generated_grid)
print("============================================")


# ------------------------------------------------------------
# 25. FINAL OBSERVATIONS
# ------------------------------------------------------------
# Use the actual outputs and plots to complete the lab report.
print("\nSuggested Observations:")
print("1. Training loss should generally decrease as the VAE learns.")
print("2. Reconstructed images should resemble the input digits.")
print("3. Generated images demonstrate sampling from the latent space.")
print("4. The latent-space plot shows how digit representations are arranged.")
print("5. Interpolation demonstrates transitions between latent representations.")
print("6. Lower MSE means lower pixel-wise reconstruction error.")
print("7. MSE alone does not completely measure visual image quality.")
print("\nExperiment 4 completed successfully.")
```
