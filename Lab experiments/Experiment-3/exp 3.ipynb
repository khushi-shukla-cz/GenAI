```python
# Experiment 3: Comparative Analysis of LLM Prompting Techniques
# Domain: Travel & Tourism | LLM used in the experiment: ChatGPT

tasks = {
    "Travel question answering": "What are the main attractions of Paris for a first-time visitor?",
    "Destination summarization": "Summarize the important things a first-time tourist should know about Paris.",
    "Travel itinerary": "Create a 3-day travel itinerary for a first-time visitor to Paris.",
    "Hotel email writing": "Write an email to a hotel requesting a room reservation for two people for three nights.",
    "Travel sentence translation": "Translate this sentence into French: 'I would like to book a room with a city view.'"
}

techniques = ["Zero-shot", "Few-shot", "Chain-of-Thought", "Role Prompting"]
```

```python
def build_prompt(task_name, task, technique):
    if technique == "Zero-shot":
        return task

    if technique == "Few-shot":
        if task_name == "Travel question answering":
            return (
                "Answer travel questions concisely and informatively.\n"
                "Example 1: Question: What are the main attractions of London? "
                "Answer: Major attractions include the London Eye, Tower of London, "
                "Buckingham Palace, Big Ben, and the British Museum.\n"
                "Example 2: Question: What are the main attractions of Rome? "
                "Answer: Major attractions include the Colosseum, Roman Forum, "
                "Pantheon, Trevi Fountain, and Vatican City.\n"
                f"Now answer: {task}"
            )
        if task_name == "Destination summarization":
            return (
                "Summarize each destination in a short paragraph covering attractions, "
                "culture, transport, and useful visitor information.\n"
                "Example 1: Kyoto — Known for temples, traditional streets, gardens, "
                "and seasonal cherry blossoms.\n"
                "Example 2: Rome — Known for ancient landmarks, museums, historic "
                "architecture, and Italian cuisine.\n"
                f"Now complete the same type of summary: {task}"
            )
        if task_name == "Travel itinerary":
            return (
                "Create itineraries in a day-wise format with morning, afternoon, "
                "and evening activities.\n"
                "Example 1: Day 1: Visit the city landmark in the morning, a museum "
                "in the afternoon, and a local market in the evening.\n"
                "Example 2: Day 1: Explore the historic centre in the morning, visit "
                "a major attraction after lunch, and enjoy a local food area at night.\n"
                f"Now create an itinerary: {task}"
            )
        if task_name == "Hotel email writing":
            return (
                "Write polite and professional hotel emails with a subject, greeting, "
                "clear request, and closing.\n"
                "Example 1: Subject: Room Availability Inquiry. Dear Team, I would "
                "like to ask about room availability for two guests. Regards, Guest.\n"
                "Example 2: Subject: Request for Late Check-in. Dear Team, Please "
                "confirm whether late check-in is possible. Regards, Guest.\n"
                f"Now write the email: {task}"
            )
        return (
            "Translate travel sentences accurately and provide only the translation.\n"
            "Example 1: 'Where is the railway station?' -> 'Où est la gare ?'\n"
            "Example 2: 'I need a taxi.' -> 'J'ai besoin d'un taxi.'\n"
            f"Now translate: {task}"
        )

    if technique == "Chain-of-Thought":
        return (
            "Complete the task carefully. Identify the key requirements, consider "
            "relevant constraints, check that the response is consistent, and then "
            "provide a clear final answer. Do not provide private internal reasoning; "
            "give only a brief explanation of the criteria used when useful.\n"
            f"Task: {task}"
        )

    if technique == "Role Prompting":
        roles = {
            "Travel question answering": "You are an experienced travel guide specializing in European tourism.",
            "Destination summarization": "You are a professional travel writer who prepares accurate destination guides.",
            "Travel itinerary": "You are a travel planner who creates practical, well-organized tourist itineraries.",
            "Hotel email writing": "You are a travel assistant experienced in professional hotel communication.",
            "Travel sentence translation": "You are a professional French translator specializing in travel communication."
        }
        return f"{roles[task_name]} Complete the following task clearly and accurately.\n{task}"

    raise ValueError("Unknown prompting technique")
```

```python
prompts = {
    task_name: {
        technique: build_prompt(task_name, task, technique)
        for technique in techniques
    }
    for task_name, task in tasks.items()
}

for task_name, technique_prompts in prompts.items():
    print("\\n" + "=" * 80)
    print(task_name.upper())
    print("=" * 80)
    for technique, prompt in technique_prompts.items():
        print(f"\\n[{technique} PROMPT]")
        print(prompt)
```

```python
# Copy each generated prompt into ChatGPT (or another LLM), then enter its response below.
# Responses are collected without assuming or fabricating model outputs.

responses = {}

for task_name in tasks:
    responses[task_name] = {}
    for technique in techniques:
        print("\\n" + "=" * 80)
        print(f"Task: {task_name} | Technique: {technique}")
        print("Prompt:\\n" + prompts[task_name][technique])
        responses[task_name][technique] = input("\\nPaste the LLM response here (or press Enter to skip): ").strip()
```

```python
# Display the collected responses
for task_name, technique_responses in responses.items():
    print("\\n" + "=" * 80)
    print(task_name.upper())
    print("=" * 80)
    for technique, response in technique_responses.items():
        print(f"\\n[{technique}]")
        print(response if response else "No response entered.")
```

```python
# Response evaluation rubric
evaluation_criteria = [
    "Response Quality",
    "Correctness",
    "Reasoning",
    "Creativity",
    "Relevance",
    "Completeness",
    "Response Length"
]

ratings = {}

for task_name in tasks:
    ratings[task_name] = {}
    for technique in techniques:
        if not responses[task_name][technique]:
            continue
        print(f"\\nRate {technique} for: {task_name}")
        ratings[task_name][technique] = {}
        for criterion in evaluation_criteria:
            ratings[task_name][technique][criterion] = input(
                f"{criterion} (e.g., Excellent/Good/Average/Poor or Short/Medium/Long): "
            ).strip()
```

```python
# Generate a comparison summary from the ratings entered above
for task_name, technique_ratings in ratings.items():
    print("\\n" + "=" * 80)
    print(task_name.upper())
    print("=" * 80)
    for technique, criterion_ratings in technique_ratings.items():
        print(f"\\n{technique}")
        for criterion, rating in criterion_ratings.items():
            print(f"{criterion}: {rating}")
```

```python
# Save prompts, responses, and ratings to a JSON file
import json

experiment_results = {
    "experiment": "Comparative Analysis of LLM Prompting Techniques in Travel & Tourism",
    "llm_used": "ChatGPT",
    "tasks": tasks,
    "prompts": prompts,
    "responses": responses,
    "ratings": ratings
}

with open("experiment_3_results.json", "w", encoding="utf-8") as file:
    json.dump(experiment_results, file, indent=4, ensure_ascii=False)

print("Experiment data saved to experiment_3_results.json")
```
